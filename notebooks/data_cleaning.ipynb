```python
import pandas as pd 
```

```python
df = pd.read_csv("telco.csv")
```

```python
df.describe(include = "all")
```

```python
df.columns = df.columns.str.lower().str.strip().str.replace(" ", "_")
```

```python
df.columns
```

```python
df.nunique()
```

```python
df = df.drop(columns = ["country","state", "quarter"])
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df["offer"] = df["offer"].fillna("No Offer")
df["internet_type"] = df["internet_type"].fillna("No Internet")
df["churn_reason"] = df["churn_reason"].fillna("Not Churned")
df["churn_category"] = df["churn_category"].fillna("Not Churned")
```

```python
df.isnull().sum()
```

```python
df = df.dropna()
```

```python
df.isnull().sum()
```

```python
df[df["senior_citizen"] == "Yes"]["age"].describe()
```

```python
df[df["senior_citizen"] == "No"]["age"].describe()
```

```python
df[df["senior_citizen"] == "Yes"]["age"].min()
```

```python
df[df["senior_citizen"] == "No"]["age"].max()
```

```python
df[df["under_30"] == "No"]["age"].min()
```

```python
df.loc[:,"age_group"] = pd.cut(
    df["age"],
    bins=[0, 29, 64, 100],
    labels=["Under 30", "Middle age", "Senior Citizen"]
)
```

```python
print(df[["age","age_group"]])
```

```python
df = df.drop(columns = ["under_30", "senior_citizen"])
```

```python
df = df.drop(columns = ["dependents", "referred_a_friend"])
```
